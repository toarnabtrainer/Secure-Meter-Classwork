## **Mango Garden Project using Python**

```python
class Tree:
    def __init__(self, code, height, width, amt):  # constructor method
        self.__code = code   # defining private instance variables
        self.__height = height
        self.__width = width
        self.__amount_spent = amt
    def set_code(self, code):   # setter method for writing
        self.__code = code
    def get_code(self):         # getter method for reading
        return self.__code
    def set_height(self, height):
        self.__height = height
    def get_height(self):
        return self.__height
    def set_width(self, width):
        self.__width = width
    def get_width(self):
        return self.__width
    def set_amount_spent(self, amt):
        self.__amount_spent = amt
    def get_amount_spent(self):
        return self.__amount_spent
    def annual_update(self, htt, wdd, amtt):
        self.__height += htt
        self.__width += wdd
        self.__amount_spent += amtt
    def display_tree(self):
        print(f"Tree Code: {self.get_code()}, Height: {self.get_height()}, Width: {self.get_width()}\
 and Amount_Spent: {self.get_amount_spent()}")
```

```python
t1 = Tree("Banyan", 40, 50, 15000)
t1.display_tree()
t1.annual_update(5, 10, 25000)
t1.display_tree()
```

```python
class MangoTree(Tree):
    def __init__(self, code, height, width, amt, yld):  # constructor method
        self.__yield = yld   # private instance variable
        super(MangoTree, self).__init__(code, width, height, amt)  # calling base class constructor
    def set_yield(self, yld):   # setter method for writing
        self.__yield = yld
    def get_yield(self):        # getter method for reading
        return self.__yield
    def annual_update(self, htt, wdd, amtt, yldd):
        super(MangoTree, self).annual_update(htt, wdd, amtt)
        self.__yield += yldd
    def display_mango_tree(self):
        super(MangoTree, self).display_tree()
        print(f"And Yield: {self.get_yield()}")
```

```python
mt1 = MangoTree("Himsagar", 20, 10, 25000, 5000)
mt1.display_mango_tree()
mt1.annual_update(10, 5, 5000, 3500)
mt1.display_mango_tree()
```

```python
class Garden:
    def __init__(self, m1, m2, m3, t1, t2):
        self.__mango1 = m1
        self.__mango2 = m2
        self.__mango3 = m3
        self.__tree1 = t1
        self.__tree2 = t2
    def total_yield(self):
        return self.__mango1.get_yield() + self.__mango2.get_yield() + self.__mango3.get_yield()
    def total_amount_spent(self):
        return self.__mango1.get_amount_spent() + self.__mango2.get_amount_spent() +\
        self.__mango3.get_amount_spent() + self.__tree1.get_amount_spent() +\
        self.__tree2.get_amount_spent()
    def garden_display(self):
        print("Please find the details about the garden below:")
        self.__mango1.display_mango_tree()
        self.__mango2.display_mango_tree()
        self.__mango3.display_mango_tree()
        self.__tree1.display_tree()
        self.__tree2.display_tree()
```

```python
grd1 = Garden(MangoTree("Himsagar", 20, 10, 25000, 5000),
              MangoTree("Langra", 22, 12, 25500, 5500),
              MangoTree("Chausa", 24, 14, 26000, 6000),
              Tree("Banyan", 40, 50, 15000),
              Tree("Palm", 45, 55, 15500)
              )
grd1.garden_display()
print(f"Total amount spent for the garden is {grd1.total_amount_spent()}...")
print(f"Total yield from the garden is {grd1.total_yield()}...")
```
