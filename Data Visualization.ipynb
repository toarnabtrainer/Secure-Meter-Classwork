```python
# rapid multiplot
x = np.linspace(0, 10, 50)
sinus = np.sin(x)
cosinus = np.cos(x)
plt.plot(x, sinus, "-b", x, sinus, "or", x, cosinus, "-r", x, cosinus, "ob")
plt.xlabel("This is X-Axis...")
plt.ylabel("This is Y-Axis...")
plt.title("Plotting my First Plot...")
plt.show()
```

```python
# step by step
x =  np.linspace(0, 10, 50)
sinus = np.sin(x)
cosinus = np.cos(x)
plt.plot(x, sinus, label='sinus', color='red', linestyle="--", linewidth=2)
plt.plot(x, cosinus, label='cosinus', color='blue', linestyle="-", linewidth=2)
plt.legend(loc=0)
plt.xlabel("This is X-Axis...")
plt.ylabel("This is Y-Axis...")
plt.title("Plotting my First Plot...")
plt.show()
```

```python
# draw a circle with origin as center and radius is 10
# x**2 + y**2 = r**2 => y = (r**2 - x**2) ** 0.5
r = 10
x = np.linspace(-10, 10, 100)
y = np.sqrt(r**2 - x**2)
plt.figure(figsize=(3, 3))
plt.axis("equal")
plt.plot(x, y, "-r")
plt.plot(x, -y, "-r")
plt.show()
```

```python
# scatter 2D plot
try:
    salary_df = pd.read_csv("https://raw.githubusercontent.com/toarnabtrainer/Secure-Meter-Classwork/refs/heads/main/Datafile/salary_table.csv")
    display(salary_df.head())
except:
    print("File access ERROR !!!")
print("End of the program...")
```

```python
print(salary_df.education.value_counts())
```

```python
df = salary_df.copy()
colors_edu = {'Master':"g", "Bachelor":"r", "Ph.D":"blue"}
plt.scatter(df['experience'], df.salary, c=df['education'].apply(lambda edu:colors_edu[edu]), s=40)
plt.title("Experience (in years) vs. Salary (in INR)")
plt.xlabel("Experience (in years)...")
plt.ylabel("Salary (in INR)...")
plt.show()
```

```python
# group by on education x management => 6 groups
for values, d in salary_df.groupby(['education', 'management']):
    print(values, type(values))
    display(d)
    print(type(d), d.shape)
```

```python
symbols_mang = {'Y':"*", 'N':"."}
colors_edu = {'Master':"g", "Bachelor":"r", "Ph.D":"blue"}
plt.figure(figsize=(6, 5))
for values, d in salary_df.groupby(['education', 'management']):
    edu, manager = values
    plt.scatter(d.experience, d['salary'], marker=symbols_mang[manager], color=colors_edu[edu], s=150, label=manager+"/"+edu)
    plt.title("Experience (in years) vs. Salary (in INR)")
    plt.xlabel("Experience (in years)...")
    plt.ylabel("Salary (in INR)...")
    plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

import matplotlib.patches as patches
import matplotlib.path as path

np.random.seed(19680801)  # Fixing random state for reproducibility

# histogram our data with numpy
data = np.random.randn(1000)
n, bins = np.histogram(data, 50)

# get the corners of the rectangles for the histogram
left = bins[:-1]
right = bins[1:]
bottom = np.zeros(len(left))
top = bottom + n

# we need a (numrects x numsides x 2) numpy array for the path helper
# function to build a compound path
XY = np.array([[left, left, right, right], [bottom, top, top, bottom]]).T

# get the Path object
barpath = path.Path.make_compound_path_from_polys(XY)

# make a patch out of it, don't add a margin at y=0
patch = patches.PathPatch(barpath)
patch.sticky_edges.y[:] = [0]

fig, ax = plt.subplots()
ax.add_patch(patch)
ax.autoscale_view()
plt.show()
```
