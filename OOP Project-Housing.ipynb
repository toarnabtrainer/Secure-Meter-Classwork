```python
class Television:
    def __init__(self, make, size, dop, isColor):
        self.__make = make
        self.__size = size
        self.__dop = dop
        self.__isColor = isColor
    def displayTelevision(self):
        print(f"Make: {self.__make}, Size: {self.__size}, DOP: {self.__dop} and IsColor: {self.__isColor}...")
```

```python
tv1 = Television("Sonodyne", "21", "20-12-2020", True)
tv1.displayTelevision()
tv2 = Television("Philips", "24", "20-02-2022", True)
tv2.displayTelevision()
```

```python
class Room:
    def __init__(self, name, length, breadth, height, fans, lights, windows, doors):
        self.__name = name
        self.__length = length
        self.__breadth = breadth
        self.__height = height
        self.__fans = fans
        self.__lights = lights
        self.__windows = windows
        self.__doors = doors
        self.__wallarea = 2 * (length + breadth) * height * 0.80
        self.__floorarea = length * breadth
    def displayRoom(self):
        print(f"Name: {self.__name}, Fans: {self.__fans}, Lights: {self.__lights}, Windows: {self.__windows} and Doors: {self.__doors}...")
        print(f"Floor Area: {self.__floorarea} and Wall Area: {self.__wallarea}...")
    def getFloorArea(self):
        return self.__floorarea
```

```python
room1 = Room("Drawing Room", 15, 10, 10, 2, 4, 3, 2)
room1.displayRoom()
print("Floor Area:", room1.getFloorArea())
room1 = Room("Study Room", 10, 8, 10, 2, 1, 2, 2)
room1.displayRoom()
print("Floor Area:", room1.getFloorArea())
```

```python
class House:
    def __init__(self, doorno, br, dr, hr, kr, dw, tv, doc):
        self.__doorno = doorno; self.__bedRoom = br
        self.__dinRoom = dr; self.__hallRoom = hr
        self.__kinRoom = kr; self.__drawRoom = dw
        self.__television = tv; self.__doc = doc
        self.__floorArea = br.getFloorArea() + dr.getFloorArea() + \
        hr.getFloorArea() + kr.getFloorArea() + dw.getFloorArea()
        self.__plinthArea = self.__floorArea * 1.50
        self.__landArea = self.__floorArea * 1.75
    def displayHouse(self):
        print(f"Door No: {self.__doorno}, DOC: {self.__doc}, Plinth Area: {self.__plinthArea} and Land Area: {self.__landArea}...")
        self.__bedRoom.displayRoom(); self.__dinRoom.displayRoom()
        self.__hallRoom.displayRoom(); self.__kinRoom.displayRoom()
        self.__drawRoom.displayRoom(); self.__television.displayTelevision()
    # OBJECTORIENTEDPROGRAAMING - Upper Case
    # objectorientedprogramming - Lower Case
    # ObjectOrientedProgramming - Proper Case
    # objectOrientedProgramming - Camel Case
```

```python
house1 = House("DN-10001",
               Room("Bed Room", 15, 10, 10, 2, 5, 4, 2),
               Room("Dining Room", 15, 10, 10, 4, 2, 2, 2),
               Room("Hall Room", 18, 11, 12, 3, 4, 3, 2),
               Room("Kitchen Room", 10, 12, 10, 1, 4, 5, 2),
               Room("Drawing Room", 15, 14, 11, 3, 4, 3, 2),
               Television("Sonodyne", "21", "20-12-2020", True), "20-12-2005")
house1.displayHouse()
```

```python
class Park:
    def __init__(self, length, breadth):
        self.__length = length
        self.__breadth = breadth
        self.__area = length * breadth
    def displayPark(self):
        print(f"Length: {self.__length}, Breadth: {self.__breadth} and Area: {self.__area}...")
```

```python
pk1 = Park(1000, 1500)
pk1.displayPark()
pk2 = Park(1500, 1800)
pk2.displayPark()
```

```python
class Block:
    def __init__(self, length, breadth, hh1, hh2, hh3, hh4, pk1):
        self.__house1 = hh1; self.__house2 = hh2
        self.__house3 = hh3; self.__house4 = hh4
        self.__park = pk1
        self.__area = length * breadth
    def displayBlock(self):
        print("Displaying the Block details...")
        print("\nDisplaying the House-1 details...")
        self.__house1.displayHouse();
        print("\nDisplaying the House-2 details...")
        self.__house1.displayHouse()
        print("\nDisplaying the House-3 details...")
        self.__house3.displayHouse();
        print("\nDisplaying the House-4 details...")
        self.__house4.displayHouse()
        print("\nDisplaying the Park details...")
        self.__park.displayPark()
        print("\nArea of the block is", self.__area)
```
